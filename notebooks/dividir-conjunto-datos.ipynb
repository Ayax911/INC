```python
from sklearn.model_selection import train_test_split
from pathlib import Path
import pandas as pd
```

```python
DATA_PROCESSED_DIR = Path("../data/processed/data.csv")     # Ruta de los datos procesados
df = pd.read_csv(DATA_PROCESSED_DIR)                        # Cargar el dataset procesado

train_df, temp_df = train_test_split(
    df, 
    test_size=0.20,                                         # 20% para test + validation
    random_state=42,                                        # Fijar semilla para reproducibilidad  
    stratify=df['Label']                                    # Mantener proporción de diagnósticos
)

test_df, val_df = train_test_split(
    temp_df,
    test_size=0.25,                                         # 25% de 20% = 5% del total para validation
    random_state=42,
    stratify=temp_df['Label'] 
)

print(f"Total de registros: {len(df)}")
print(f"\nTrain: {len(train_df)} registros ({len(train_df)/len(df)*100:.1f}%)")
print(f"Test:  {len(test_df)} registros ({len(test_df)/len(df)*100:.1f}%)")
print(f"Val:   {len(val_df)} registros ({len(val_df)/len(df)*100:.1f}%)")

print(f"\nDistribución de diagnósticos en Train:")
print(train_df['Label'].value_counts(normalize=True))
print(f"\nDistribución de diagnósticos en Test:")
print(test_df['Label'].value_counts(normalize=True))
print(f"\nDistribución de diagnósticos en Validation:")
print(val_df['Label'].value_counts(normalize=True))
```

```python
import os

output_dir = '../data/processed/parches_224x224/splits/'            # Directorio de salida
os.makedirs(output_dir, exist_ok=True)                              # Crear el directorio si no existe

# Guardar cada conjunto
train_path = os.path.join(output_dir, 'train_clinical_data.csv')    # Ruta del archivo de entrenamiento
test_path = os.path.join(output_dir, 'test_clinical_data.csv')      # Ruta del archivo de prueba
val_path = os.path.join(output_dir, 'val_clinical_data.csv')        # Ruta del archivo de validación

train_df.to_csv(train_path, index=False)                            # Guardar conjunto de entrenamiento
test_df.to_csv(test_path, index=False)                              # Guardar conjunto de prueba
val_df.to_csv(val_path, index=False)                                # Guardar conjunto de validación

print("✓ Archivos guardados exitosamente:")
print(f"  - Train: {train_path}")
print(f"  - Test:  {test_path}")
print(f"  - Val:   {val_path}")
```
